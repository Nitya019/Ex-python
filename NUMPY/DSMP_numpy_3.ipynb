###`Q-2:` There is an array of marks of 5 students in 4 subjects. Further you are asked to perform below task.
    i. Add marks every student of an extra subject in the same array.
    ii. Add two new students marks in respective 5 subjects.(one subject added in above task)
    iii. Add extra column with sum of all subjects(5-subjects) marks
    iv. Sort the array(non-ascending order) on total marks column--one added in above task. Show top 2 rows.

Note: Change dimension of arrays during concatenation or appending if required.

Given Array-
```
marks = [[13, 10,  9, 33],
       [63, 46, 90, 42],
       [39, 76, 13, 29],
       [82,  9, 29, 78],
       [67, 61, 59, 36]]

extra_subject = [41, 87, 72, 36, 92]
#Two extra students record-
rec1 = [77, 83, 98, 95, 89]
rec2 = [92, 71, 52, 61, 53]
```

````python
marks = np.array([13,10,9,33,63,46,90,42,39,76,13,29,82,9,29,78,67,61,59,36]).reshape(5,4)
extra_subject = np.array([41,87, 72,36,92]).reshape(5,1)
rec1 = np.array([77, 83, 98, 95, 89]).reshape(1,5)
rec2 = np.array([92, 71, 52, 61, 53]).reshape(1,5)

extra_subject.ndim
````

````python
#1)
a = np.hstack((marks, extra_subject))
a
````

````python
#2) adding two more students

flat_rec = np.ravel(a)
new = np.append(a, rec1)
final = np.append(new, rec2)

final.reshape(7,5)
````

````python
#3)add extra columns with sum of all subjects
final = np.array(final).reshape(7,5)

row_sums = final.sum(axis=1)
final_with_sum = np.column_stack((final, row_sums))

final_with_sum
````

````python
#4) 

row_sums = np.sort(final_with_sum)[::-1]

row_sums
````

### `Q-7:` You are given a array. You have to find the minimum and maximum array element and remove that from the array.

```python
import numpy as np

np.random.seed(400)
arr = np.random.randint(100, 1000, 200).reshape((1, 200))
```

````python
import numpy as np

np.random.seed(400)
arr = np.random.randint(100, 1000, 200).reshape((1, 200))
````

````python
print(np.argmax(arr))
print(np.argmin(arr))
````

````python
np.delete(arr, np.where(np.argmax(arr)))
np.delete(arr, np.where(np.argmin(arr)))
````
